# IRR

```python
merged_df = pd.read_csv("/content/merged_annotations.csv")
```

```python
merged_df
```

```python
# Extract label columns
annotations = merged_df[["lbl1", "lbl2", "lbl3"]]
```

```python
# Encode string labels into integers
label_mapping = {label: idx for idx, label in enumerate(pd.unique(annotations.values.ravel()))}
encoded_annotations = annotations.replace(label_mapping)
```

Krippendorff’s Alpha

```python
# Transpose to shape: raters x items
kripp_alpha_input = encoded_annotations.T.values
```

```python
# Compute Krippendorff's Alpha (nominal level)
kripp_alpha = krippendorff.alpha(reliability_data=kripp_alpha_input, level_of_measurement='nominal')
print("Krippendorff's Alpha:", kripp_alpha)
```

Fleiss’ Kappa

```python
# Count number of times each category was chosen per item
fleiss_input = encoded_annotations.apply(lambda row: row.value_counts(), axis=1).fillna(0)
```

```python
# Ensure consistent order of columns (i.e., categories)
for col in range(len(label_mapping)):
    if col not in fleiss_input.columns:
        fleiss_input[col] = 0
fleiss_input = fleiss_input[sorted(fleiss_input.columns)]
```

```python
# Convert to matrix
fleiss_matrix = fleiss_input.values.astype(int)
```

```python
# Compute Fleiss' Kappa
fleiss_score = fleiss_kappa(fleiss_matrix)
print("Fleiss' Kappa:", fleiss_score)
```

**Krippendorff's Alpha:** *0.371*

**Fleiss' Kappa:** *0.371*

3 annotators had a fair agreement

# Final Label

```python
# Function to compute majority vote for each row
def majority_vote(row):
    return row.mode().iloc[0]  # picks the most common label
```

```python
# Assign final label
merged_df["fin_lbl"] = merged_df[["lbl1", "lbl2", "lbl3"]].apply(majority_vote, axis=1)
```

```python
duplicated_count = duplicated_entries['userID'].value_counts()

print(duplicated_count)
```

Merge votes from all duplicates, then recompute majority vote per userID

```python
# Group by userID and gather all annotations
grouped = duplicated_entries.groupby('userID').agg({
    'lbl1': lambda x: list(x),
    'lbl2': lambda x: list(x),
    'lbl3': lambda x: list(x),
})
```

```python
# Flatten the labels into a single list per userID
grouped['all_labels'] = grouped.apply(lambda row: row['lbl1'] + row['lbl2'] + row['lbl3'], axis=1)
```

```python
# Majority vote again
def true_majority(labels):
    return pd.Series(labels).mode().iloc[0]  # Most frequent label
```

```python
grouped['recomputed_final_label'] = grouped['all_labels'].apply(true_majority)
```

```python
print(grouped[['all_labels', 'recomputed_final_label']])
```

```python
duplicated_check = grouped.index.duplicated().sum()

print(f"Number of duplicated userIDs after recomputing: {duplicated_check}")
```
